```python
# Stack Implementation with Dynamic Array
```

```python
# Initiate an array
stack = []
```

```python
# Push
stack.append('A')
stack.append('B')
stack.append('C')
print('Stack:',stack)
```

```python
# Peek
topelement = stack[-1]
print(topelement)
```

```python
# Pop
popped = stack.pop(1)
print(popped)
```

```python
# isEmpty
x = bool(stack)
print(x)

if(x== True):
  print('stack is not empty')
```
