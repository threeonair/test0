```python
# Stack with Class
"""
We will try to create a stack class and use it
"""
```

```python
class Stack:
  def __init__(self):
    self.stack=[]

  def push(self,element):
    self.stack.append(element)

  def pop(self):
    if self.isEmpty():
      return "Stack is empty:"
    return self.stack.pop()

  def peek(self):
    if self.isEmpty():
      return "Stack is empty"
    return self.stack[-1]


  def isEmpty(self):
    return len(self.stack)==0

  def size(self):
    return len(self.stack)
```

```python
# Create a Stadck
myStack = Stack()
```

```python
print(myStack.stack)
print(myStack.isEmpty)
```

```python
print(myStack.pop())
```

```python
print(myStack.pop)
```

```python
print(myStack.isEmpty)
```

```python
print(myStack.size())
```

```python
print(myStack.peek())
```
